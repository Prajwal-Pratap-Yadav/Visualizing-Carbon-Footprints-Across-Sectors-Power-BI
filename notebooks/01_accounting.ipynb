```python
from pathlib import Path
import json
from decimal import Decimal
from carbon_audit.analytics import percent_change
ROOT = Path.cwd()
if not (ROOT / 'configs/policy.json').exists(): ROOT = ROOT.parent
profile = json.loads((ROOT/'reports/reconciliation.json').read_text())
world = json.loads((ROOT/'reports/analysis.json').read_text())['WORLD']
print('Source SHA256:', profile['source_sha256'])
print('Database licence:', profile['data_license'])
```

```python
print('Rows retained:', profile['retained_rows'])
print('Calendar days:', profile['calendar_days'])
print('Geography labels:', len(profile['geographies']))
print('Sector labels:', len(profile['sectors']))
print('Coverage:', profile['first_day'], 'to', profile['last_day'])
```

```python
print('All-label sum / WORLD:', profile['all_labels_sum_divided_by_world'])
print('Disjoint partition:', ', '.join(profile['partition']))
print('Independent checks:', profile['spatial_checks'], 'passes:', profile['spatial_passes'])
print('Max absolute residual, interpreted MtCO2:', profile['maximum_absolute_spatial_residual_mt_co2'])
print('No independent all-sector total:', profile['all_sector_total_status'])
```

```python
for p in world['periods']:
    print(p['year'], p['period_kind'], p['days'], p['derived_six_sector_total_mt_co2'], 'YoY%:', p['yoy_percent_complete_years_only'])
print('CAGR 2019–2022 fraction:', world['cagr_complete_years_fraction'])
```

```python
for p in world['matched_ytd']:
    print(p['year'], p['first_day'], p['last_day'], p['calendar_days'], p['derived_six_sector_total_mt_co2'])
current, previous = world['matched_ytd'][-1], world['matched_ytd'][-2]
print('2023 versus 2022 Jan–May change%:', percent_change(Decimal(current['derived_six_sector_total_mt_co2']), Decimal(previous['derived_six_sector_total_mt_co2'])))
```

```python
period = next(p for p in world['periods'] if p['year']==2022)
for sector, share in period['sector_shares_percent'].items():
    print(sector, 'share of WORLD six-sector 2022 total%:', share)
print('Population supplied:', profile['population'])
print('Scientific accuracy certified:', profile['scientific_accuracy_or_primary_inventory_certified'])
```
